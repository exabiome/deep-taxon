```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np
import os
```

```python
gtnet_ctg_path = os.path.join(os.environ["SCRATCH"], "exabiome/deep-taxon/gtnet/sota/contigs/results.gtnet.npz")
cat_ctg_path = os.path.join(os.environ["SCRATCH"], "exabiome/deep-taxon/gtnet/sota/contigs/results.cat.npz")
sourmash_ctg_path = os.path.join(os.environ["SCRATCH"], "exabiome/deep-taxon/gtnet/sota/contigs/results.sourmash.npz")
```

```python
def read(path):
    results = np.load(path, allow_pickle=True)
    results = {k: results[k] for k in results.keys()}
    results['complete'] = results['complete'].item()
    results['correct'] = results['correct'].item()
    return results

gtnet_results = read(gtnet_ctg_path)
cat_results = read(cat_ctg_path)
sourmash_results = read(sourmash_ctg_path)
```

```python
def check_ax(ax):
    if ax is None:
        return plt.gca()
    return ax

def _plot_helper(func, tax_level, title, ax=None):
    gc = 'g'
    sc = 'b'
    cc = 'r'
    slc = 'gray'
    
    ax = check_ax(ax)
    x = (gtnet_results['bins'][1:] + gtnet_results['bins'][:-1]) / 2
    ax.plot(x, func(cat_results, tax_level), c=cc)
    ax.plot(x, func(sourmash_results, tax_level), c=sc)
    ax.plot(x, func(gtnet_results, tax_level), c=gc)

    ax.set_ylabel(title)
    ax.set_xlabel("Sequence length")
    ax.set_xscale('log')
    len_ax = ax.twinx()
    len_ax.plot(x, gtnet_results['total'] / gtnet_results['total'].sum(), c=slc, ls='--')
    len_ax.set_ylabel("Frequency (sequence length)")

    custom_lines = [Line2D([0], [0], color=gc),
                    Line2D([0], [0], color=sc),
                    Line2D([0], [0], color=cc),
                    Line2D([0], [0], color='gray', ls='--')]
    ax.legend(custom_lines, ['GTNet', 'Sourmash', 'CAT', 'seq. len.'])
    
    return ax

    
def calc_acc(results, tax_level):
    result = results['correct'][tax_level] / results['complete'][tax_level]
    result[np.isnan(result)] = 0.0
    return result


def calc_comp(results, tax_level):
    result = results['complete'][tax_level] / results['total']
    result[np.isnan(result)] = 0.0
    return result


def plot_complete(tax_level, ax=None):    
    ax = _plot_helper(calc_comp, tax_level, "Classification rate", ax=ax)

    
def plot_accuracy(tax_level, ax=None):
    ax = _plot_helper(calc_acc, tax_level, "Accuracy", ax=ax)
```

```python
def _plot_all_helper(func, tax_level, ax=None, lylabel=True, xlabel=True):
    gc = 'g'
    sc = 'b'
    cc = 'r'
    slc = 'gray'
    
    ax = check_ax(ax)
    x = (gtnet_results['bins'][1:] + gtnet_results['bins'][:-1]) / 2
    ax.plot(x, func(cat_results, tax_level), c=cc)
    ax.plot(x, func(sourmash_results, tax_level), c=sc)
    ax.plot(x, func(gtnet_results, tax_level), c=gc)


    ax.set_xscale('log')
    len_ax = ax.twinx()
    len_ax.plot(x, gtnet_results['total'] / gtnet_results['total'].sum(), c=slc, ls='--')

    if lylabel:
        len_ax.set_ylabel("Frequency (sequence length)", fontsize=14)
    else:
        len_ax.set_yticklabels([])

    if xlabel:
        ax.set_xlabel("Sequence length", fontsize=14)
    

    custom_lines = [Line2D([0], [0], color=gc),
                    Line2D([0], [0], color=sc),
                    Line2D([0], [0], color=cc),
                    Line2D([0], [0], color='gray', ls='--')]
    ax.legend(custom_lines, ['GTNet', 'Sourmash', 'CAT', 'seq. len.'])
    
    return ax

    
def calc_acc(results, tax_level):
    result = results['correct'][tax_level] / results['complete'][tax_level]
    result[np.isnan(result)] = 0.0
    return result


def calc_comp(results, tax_level):
    result = results['complete'][tax_level] / results['total']
    result[np.isnan(result)] = 0.0
    return result
```

```python
fig, axes = plt.subplots(7, 2, figsize=plt.gcf().get_size_inches() * [2, 5], sharey=True, sharex=True)
for k, _axes in zip(gtnet_results['complete'], axes):
    _plot_all_helper(calc_comp, k, ax=_axes[0], lylabel=False, xlabel=k=='species')
    _plot_all_helper(calc_acc, k, ax=_axes[1], xlabel=k=='species')
    _axes[0].set_ylabel(k, rotation=0, fontsize=14)
    _axes[0].yaxis.set_label_coords(-0.15, 0.5)

axes[0, 0].set_title("Classification rate", fontsize=18)
axes[0, 1].set_title("Accuracy", fontsize=18)
plt.tight_layout();
plt.savefig("PlotPredPerf/contigs_all.png", dpi=100)
```

```python
import pandas as pd
total = gtnet_results['total'].sum()
def get_df(results): #, weighted=False):
    data = list()
    for i, lvl  in enumerate(results['complete'].keys()):
        # if weighted:
        #     weights = results['total_bases'] / results['total_bases'].sum()
        #     complete = (weights * results['complete'][lvl]).sum()
        #     correct = (weights * results['correct'][lvl]).sum()
        # else:
        #     complete = results['complete'][lvl].sum()
        #     correct = results['correct'][lvl].sum()
        complete = results['complete'][lvl].sum()
        correct = results['correct'][lvl].sum()

        data.append({'level': lvl, 'correct': correct, 'complete': complete})
    df = pd.DataFrame(data=data).set_index('level')
    df['incorrect'] = df['complete'] - df['correct']
    return df

df = get_df(gtnet_results)
df
```

```python
def add_bars(_ax, _df, offset, colors):
    xpos = np.arange(len(df)) + offset
    bottom = np.zeros(len(df))
    for i, col in enumerate(('correct', 'incorrect')):
        _ax.bar(xpos, _df[col], width=0.25, bottom=bottom, color=colors[i])
        bottom += _df[col]
```

```python
fig, ax = plt.subplots()

cmap = plt.get_cmap('tab20')
colors = list(map(cmap, range(6)))
    
for j, results in enumerate((gtnet_results, cat_results, sourmash_results)):
    df = get_df(results)
    add_bars(ax, df, 0.25 * (j-1), colors[j*2:(j+1)*2])

ax.set_ylim(top=total)
ax.set_xticks(np.arange(7), df.index)

from matplotlib.patches import Patch

legend_elements = [Patch(facecolor=colors[0], label='GTNet'),
                   Patch(facecolor=colors[2], label='CAT'),
                   Patch(facecolor=colors[4], label='Sourmash'),
                   Patch(facecolor='k', label='Correct'),
                   Patch(facecolor=cmap(15), label='Incorrect')]

ax.legend(handles=legend_elements, loc='upper right')

ax.set_ylabel("Number of contigs")
ax.set_xlabel("Taxonomic level")

plt.savefig('PlotPredPerf/contigs_bar_unweighted.png')
```
